We apply the Apriori algorithm to Portugal's online retail data.

##DATASET DESCRIPTION
The Portugal dataset represents online purchases that were made by customers based in Portugal. The dataset size has 58 transactions for 713 products indicating that most products appear in fewer transactions. A look at the dataset indicates that the dataset in already pre-processed.

```python
import warnings
warnings.filterwarnings('ignore')
import warnings
import numpy as np
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# Load the Portugal basket dataset and remove the transaction ID column

basket_Portugal = pd.read_csv('Portugal_online_retail.csv')
basket_Portugal = basket_Portugal.drop('InvoiceNo', axis=1)
basket_Portugal
```

```python
# Find frequent itemsets — only keep itemsets appearing in >= 10% of transactions
frq_items = apriori(basket_Portugal, min_support=0.1, use_colnames=True)
frq_items.sort_values(by=['support'], ascending=False)
```

```python
# --- Confidence Level 1: Low (0.3) ---
# Lower threshold — returns more rules, including weaker associations
rules_30 = association_rules(frq_items, metric='confidence', min_threshold=0.3)
print(f'min_confidence=0.3 → {len(rules_30)} rules')
rules_30.sort_values(by=['lift'], ascending=False)
```

```python
# --- Confidence Level 2: Medium (0.6) ---
# Balanced threshold — filters out weaker rules while keeping moderate associations
rules_60 = association_rules(frq_items, metric='confidence', min_threshold=0.6)
print(f'min_confidence=0.6 → {len(rules_60)} rules')
rules_60.sort_values(by=['lift'], ascending=False)
```

```python
# --- Confidence Level 3: High (0.9) --- SELECTED
# Strictest threshold — only retains rules where the antecedent implies
# the consequent in at least 90% of transactions, removing noise


rules_90 = association_rules(frq_items, metric='confidence', min_threshold=0.9)
print(f'min_confidence=0.9 → {len(rules_90)} rules  ← SELECTED')
rules_90.sort_values(by=['lift'], ascending=False)
```

```python
# Top 3 most important rules for Portugal, ranked by lift
# Lift > 1 means the items are more likely to be bought together than by chance
top3 = rules_90.sort_values(by=['lift'], ascending=False).head(3).reset_index(drop=True)
top3
```

## Report — Portugal

**Confidence level selected: 0.9**

With 58 transactions and min_support set to 0.1, three confidence thresholds were tested on the dataset:
- **0.3** — characterised by a large number of rules. Many of the rules were identified as weak and unlikely to show genuine purchasing behaviour.
- **0.6** — indicates a more balanced set of rules.It however includes associations that are unreliable enough to be acted on commercially.
- **0.9** — This threshold was selected as it produces the most actionable and trustworthy  insights with the buying pattern holding in at least 90% of the cases. Retains only the strongest rules.
---

**Rule 1: {SCANDINAVIAN PAISLEY PICNIC BAG} → {PINK VINTAGE PAISLEY PICNIC BAG}**  
Support = 0.1207, Confidence = 1.0, Lift = 7.25
Customers who bought the Scandinavian Paisley Picnic Bag also bought the Pink Vintage Paisley Picnic Bag with no exception. The lift of 7.25 means this co-purchase occurs 7.25 times more likely than if the two products were independently bought. The two items are effectively  purchased as a set, and the retailer should put them together.

---

**Rule 2: {PLASTERS IN TIN CIRCUS PARADE, PLASTERS IN TIN VINTAGE PAISLEY} → {PLASTERS IN TIN WOODLAND ANIMALS}**  
Support = 0.1034, Confidence = 1.0, Lift = 7.25

Customers who bought, both the Circus Parade and Vintage Paisley plaster tins also bought the Woodland Animals tin. This implies Portuguese customers tend to collect the full Plasters in Tin range instead of buying a single design. For this kind of market,a bundled three-pack offer or a 'complete the collection' prompt would be a great and effective upselling strategy.

---

**Rule 3: {SCANDINAVIAN PAISLEY PICNIC BAG, LUNCH BAG RED RETROSPOT} → {PINK VINTAGE PAISLEY PICNIC BAG}**  
Support = 0.1034, Confidence = 1.0, Lift = 7.25

Portuguese customers who had both the Scandinavian Paisley Picnic Bag and the Red Retrospot Lunch Bag in their basket also added the Pink Vintage Paisley Picnic bag. This extends Rule 1 and reveals a distinct Portuguese customer segment that gravitates toward coordinated paisley-patterned bag sets. Recommending the Pink Vintage Paisley bag whenever any combination of these items are in in the cart would be appropriate.
